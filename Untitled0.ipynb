```python
import yfinance as yf

def analizar(inicio):
    datos = yf.download("MXN=X", start=inicio, progress=False)
    cierre = datos["Close"].squeeze()
    rend = cierre.pct_change().dropna()
    vol_anual = rend.std() * (252 ** 0.5)
    rend_anual = rend.mean() * 252
    return float(vol_anual), float(rend_anual), float(rend_anual / vol_anual)

print(analizar("2024-01-01"))
```

```python
ventanas = ["2022-01-01", "2023-01-01", "2024-01-01", "2025-01-01", "2026-01-01"]
print(ventanas[0])
print(len(ventanas))
```

```python
for inicio in ventanas:
    vol, rend, ratio = analizar(inicio)
    print(inicio, "| Vol:", round(vol, 3), "| Rend:", round(rend, 3), "| Ratio:", round(ratio, 2))
```

```python
def analizar(inicio, tasa_libre=0.4):
    datos = yf.download("MXN=X", start=inicio, progress=False)
    cierre = datos["Close"].squeeze()
    rend = cierre.pct_change().dropna()
    vol_anual = rend.std() * (252 ** 0.5)
    rend_anual = rend.mean() * 252
    sharpe = (rend_anual - tasa_libre) / vol_anual
    return float(vol_anual), float(rend_anual), float(sharpe)
```

```python
import yfinance as yf

def analizar(inicio, tasa_libre=0.0):
    datos = yf.download("MXN=X", start=inicio, progress=False)
    cierre = datos["Close"].squeeze()
    rend = cierre.pct_change().dropna()
    vol_anual = rend.std() * (252 ** 0.5)
    rend_anual = rend.mean() * 252
    sharpe = (rend_anual - tasa_libre) / vol_anual
    return float(vol_anual), float(rend_anual), float(sharpe)
```

```python
analizar("2024-01-01", 0.04)
```

```python
vol, rend, sharpe = analizar("2024-01-01", 0.04)
print("Vol:", round(vol, 3), "| Rend:", round(rend, 3), "| Sharpe:", round(sharpe, 2))
```
