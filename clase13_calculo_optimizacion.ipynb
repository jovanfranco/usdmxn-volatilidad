```python
import numpy as np
import yfinance as yf

# 1. Derivada con acercamientos
f = lambda x: x**2
for h in [1, 0.1, 0.01, 0.001]:
    print("h =", h, "| pendiente en x=5:", round((f(5 + h) - f(5)) / h, 4))

# 2. Portafolio de varianza mínima, dos activos sin correlación
w = np.linspace(0, 1, 1001)
var = w**2 * 0.10**2 + (1 - w)**2 * 0.20**2
print("Peso óptimo en el activo 1:", round(float(w[var.argmin()]), 3),
      "| Volatilidad:", round(float(np.sqrt(var.min())), 4))

# 3. Mínimos cuadrados con los datos del ejemplo
x = np.array([1, 2, 3.0])
y = np.array([2, 3, 5.0])
print("Pendiente por el origen:", round(float((x * y).sum() / (x**2).sum()), 4))
pend, inter = np.polyfit(x, y, 1)
print("Con constante:", round(float(pend), 4), "y", round(float(inter), 4))

# 4. Gradiente descendente en f(x) = x^2 - 4x + 5
x0 = 0.0
for paso in range(1, 9):
    x0 = x0 - 0.25 * (2 * x0 - 4)
    print("Paso", paso, "| x =", round(x0, 4))

# 5. Con datos reales: pendiente de USD/MXN contra USD/JPY
datos = yf.download(["MXN=X", "JPY=X"], start="2020-10-07", end="2026-10-22",
                    progress=False)["Close"]
rend = datos.pct_change().dropna()
beta, alfa = np.polyfit(rend["JPY=X"], rend["MXN=X"], 1)
print("Pendiente de USD/MXN contra USD/JPY:", round(float(beta), 3))
print("Correlación:", round(float(rend["JPY=X"].corr(rend["MXN=X"])), 3))
```
