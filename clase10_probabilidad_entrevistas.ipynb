```python
import numpy as np

rng = np.random.default_rng(42)

def lanzar_hasta(patron, n=20000):
    totales = []
    for _ in range(n):
        ultimos = ""
        conteo = 0
        while not ultimos.endswith(patron):
            ultimos += "H" if rng.random() < 0.5 else "T"
            conteo += 1
        totales.append(conteo)
    return np.mean(totales)

def llega_a_meta(inicio, meta, p, n=20000):
    exitos = 0
    for _ in range(n):
        capital = inicio
        while 0 < capital < meta:
            capital += 1 if rng.random() < p else -1
        exitos += capital == meta
    return exitos / n

def cumpleanos(personas, n=20000):
    repetidos = 0
    for _ in range(n):
        fechas = rng.integers(0, 365, size=personas)
        repetidos += len(set(fechas)) < personas
    return repetidos / n

def monty(cambiar, n=20000):
    gana = 0
    for _ in range(n):
        premio = rng.integers(3)
        eleccion = rng.integers(3)
        opciones = [d for d in range(3) if d != eleccion and d != premio]
        abre = opciones[rng.integers(len(opciones))]
        if cambiar:
            eleccion = [d for d in range(3) if d != eleccion and d != abre][0]
        gana += eleccion == premio
    return gana / n

def tiradas_todas_las_caras(n=20000):
    total = 0
    for _ in range(n):
        vistas = set()
        cuenta = 0
        while len(vistas) < 6:
            vistas.add(int(rng.integers(1, 7)))
            cuenta += 1
        total += cuenta
    return total / n

print("Hasta HH:", round(lanzar_hasta("HH"), 2))
print("Hasta HT:", round(lanzar_hasta("HT"), 2))
print("Ruina justo, 10 a 20:", round(llega_a_meta(10, 20, 0.5), 3))
print("Ruina con 55%, 10 a 20:", round(llega_a_meta(10, 20, 0.55), 3))
print("Cumpleaños, 23 personas:", round(cumpleanos(23), 3))
print("Monty Hall, sin cambiar:", round(monty(False), 3))
print("Monty Hall, cambiando:", round(monty(True), 3))
print("Todas las caras del dado:", round(tiradas_todas_las_caras(), 2))
```
