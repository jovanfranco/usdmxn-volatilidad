```python
import numpy as np

rng = np.random.default_rng(42)
tiradas = rng.integers(1, 7, size=100000)

print("Promedio:", round(float(tiradas.mean()), 3))
print("Varianza:", round(float(tiradas.var()), 3))
print("Desviación estándar:", round(float(tiradas.std()), 3))
```

```python
import numpy as np

rng = np.random.default_rng(42)
tiradas = rng.integers(1, 7, size=100)

print("Promedio:", round(float(tiradas.mean()), 3))
print("Varianza:", round(float(tiradas.var()), 3))
print("Desviación estándar:", round(float(tiradas.std()), 3))
```

```python
apuestas = rng.choice([1, -1], size=(100000, 100), p=[0.55, 0.45])
totales = apuestas.sum(axis=1)
print("Ganancia promedio en 100 apuestas:", round(float(totales.mean()), 2))
print("Desviación estándar:", round(float(totales.std()), 2))
print("Proporción de jugadores en pérdida:", round(float((totales < 0).mean()), 3))
```

```python
import numpy as np

rng = np.random.default_rng(42)

# Dado
tiradas = rng.integers(1, 7, size=100000)
print("Dado | Promedio:", round(float(tiradas.mean()), 3), "| Varianza:", round(float(tiradas.var()), 3))

# Apuesta con ventaja
apuestas = rng.choice([1, -1], size=(100000, 100), p=[0.55, 0.45])
totales = apuestas.sum(axis=1)
print("100 apuestas | Promedio:", round(float(totales.mean()), 2), "| Desv. estándar:", round(float(totales.std()), 2))
print("Proporción en pérdida:", round(float((totales < 0).mean()), 3))
```
