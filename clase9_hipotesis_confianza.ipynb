```python
import numpy as np
import yfinance as yf
from scipy import stats

# Paso 2: p-valor exacto con la binomial
print("60 de 100:", round(float(stats.binomtest(60, 100, 0.5, alternative="greater").pvalue), 4))
print("55 de 100:", round(float(stats.binomtest(55, 100, 0.5, alternative="greater").pvalue), 4))

# Pasos 3 y 4: intervalo y estadístico t con USD/MXN
datos = yf.download("MXN=X", start="2020-10-07", end="2026-10-22", progress=False)
cierre = datos["Close"].squeeze()
rend = cierre.pct_change().dropna()

n = len(rend)
media = float(rend.mean())
error = float(rend.std() / np.sqrt(n))
print("Media diaria:", round(media, 6), "| Error estándar:", round(error, 6))
print("Intervalo 95%:", round(media - 1.96 * error, 6), "a", round(media + 1.96 * error, 6))
print("Estadístico t:", round(media / error, 2))

# Probabilidad de al menos un falso positivo en 20 pruebas
print("Falso positivo en 20 pruebas:", round(1 - 0.95 ** 20, 3))
```
