```python
import numpy as np

# 1. Rendimiento del portafolio: matriz por vector (los 3 días de hoy)
w = np.array([0.5, 0.3, 0.2])
R = np.array([[0.02, 0.01, -0.01],
              [-0.01, 0.02, 0.00],
              [0.01, -0.02, 0.03]])
print("Rendimientos del portafolio:", R @ w)

# 2. Varianza de dos activos con la matriz de covarianza
w2 = np.array([0.5, 0.5])
cov = np.array([[0.01, 0.005],
                [0.005, 0.01]])
var = w2 @ cov @ w2
print("Varianza:", round(float(var), 4), "| Volatilidad:", round(float(np.sqrt(var)), 4))
```

```python
import yfinance as yf
import pandas as pd

datos = yf.download(["MXN=X", "JPY=X", "EURUSD=X"],
                    start="2020-10-07", end="2026-10-22", progress=False)["Close"]
datos["EURUSD=X"] = 1 / datos["EURUSD=X"]
rend = datos.pct_change().dropna()

cov_anual = rend.cov() * 252
print("Volatilidad anual de cada par:")
print((rend.std() * np.sqrt(252)).round(3))
print("Correlaciones:")
print(rend.corr().round(2))

w3 = np.array([1/3, 1/3, 1/3])
var_p = w3 @ cov_anual.values @ w3
print("Volatilidad del portafolio igual:", round(float(np.sqrt(var_p)), 4))
print("Promedio simple de las volatilidades:", round(float((rend.std() * np.sqrt(252)).mean()), 4))
```
