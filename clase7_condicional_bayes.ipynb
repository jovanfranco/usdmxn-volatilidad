```python
import numpy as np

dias = 100
noticia = np.array([1] * 20 + [0] * 80)
mov = np.array([1] * 10 + [0] * 10 + [1] * 8 + [0] * 72)

print("P(mov):", mov.mean())
print("P(mov | noticia):", mov[noticia == 1].mean())
print("P(mov | no noticia):", mov[noticia == 0].mean())
print("P(noticia | mov):", noticia[mov == 1].mean())
```
