```python
import yfinance as yf

datos = yf.download("MXN=X", start="2026-01-01")
cierre = datos["Close"]
print(cierre.tail())
```

```python
rend = cierre.pct_change().dropna()
print("Rendimiento diario promedio:", rend.mean())
print("Volatilidad diaria:", rend.std())
print("Volatilidad anualizada:", rend.std() * (252 ** 0.5))
```

```python
import yfinance as yf

datos = yf.download("MXN=X", start="2024-01-01")
cierre = datos["Close"].squeeze()

rend = cierre.pct_change().dropna()
vol_diaria = rend.std()
vol_anual = vol_diaria * (252 ** 0.5)
rend_anual = rend.mean() * 252

print("Volatilidad diaria:", float(vol_diaria))
print("Volatilidad anualizada:", float(vol_anual))
print("Rendimiento anualizado:", float(rend_anual))
print("Rendimiento entre volatilidad:", float(rend_anual / vol_anual))
```
