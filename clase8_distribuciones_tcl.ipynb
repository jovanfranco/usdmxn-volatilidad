```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

rng = np.random.default_rng(42)

# 1. Binomial
n, p = 10, 0.55
print("P(6 ganadoras):", round(float(stats.binom.pmf(6, n, p)), 4))
print("P(4 o menos):", round(float(stats.binom.cdf(4, n, p)), 4))

# 2. Ley de los grandes números
tiradas = rng.integers(1, 7, size=10000)
prom_acum = tiradas.cumsum() / np.arange(1, 10001)

plt.figure(figsize=(9, 3))
plt.plot(prom_acum)
plt.axhline(3.5, color="red", linestyle="--", label="Esperanza 3.5")
plt.title("Promedio acumulado de un dado")
plt.legend()
plt.show()

# 3. Teorema central del límite
promedios = rng.integers(1, 7, size=(20000, 30)).mean(axis=1)
x = np.linspace(promedios.min(), promedios.max(), 200)

plt.figure(figsize=(9, 3))
plt.hist(promedios, bins=40, density=True, alpha=0.6, label="Promedios de 30 tiradas")
plt.plot(x, stats.norm.pdf(x, 3.5, 1.708 / np.sqrt(30)), color="red", label="Normal")
plt.legend()
plt.show()
```
